#### Domain –Education Focus –Data analysis

#### You  are  a  data  analyst  with the University  of  California.The University  has  data  of Maths, Physics and Data Structure score of sophomore students. This data is stored in different files. The University has hired a data science company to do analysis of scores and find if there is any correlation of score with age, ethnicity,etc. Before the data is given to the company,you are required to do data wrangling.

1. Read the three CSVfiles,which contain the score ofthesame students in term1 for each Subject
2. Remove the name and ethnicity column (to ensure confidentiality)
3. Fill missing score data with zero
4. Merge the three files
5. Change Sex(M/F) Column to 1/2 for further analysis
6. Store the data inanew file –ScoreFinal.csv
7. You can try these enhancements in code
    1. Convert ethnicity to a numerical value
    2. Fill the missing score for a student to the average of the class

```python
import pandas as pd
import numpy as np
math_score_df = pd.read_csv('../requirement/MathScoreTerm1.csv')
physics_score_df = pd.read_csv('../requirement/PhysicsScoreTerm1.csv')
ds_score_df = pd.read_csv('../requirement/DSScoreTerm1.csv')

physics_score_df.info()
math_score_df.info()
ds_score_df.info()
```

```python
physics_score_df.drop(['Name','Ethinicity'], axis=1, inplace=True)
ds_score_df.drop(['Name','Ethinicity'], axis=1, inplace=True)
math_score_df.drop(['Name','Ethinicity'], axis=1, inplace=True)
```

```python
physics_score_df['Score'] = physics_score_df['Score'].replace(to_replace=np.nan, value=0)
physics_score_df.rename(columns={'Score':'Physics'}, inplace=True)
physics_score_df.drop(['Subject'], axis=1, inplace=True)
math_score_df['Score'] = math_score_df['Score'].replace(to_replace=np.nan, value=0)
math_score_df.rename(columns={'Score':'Maths'}, inplace=True)
math_score_df.drop(['Subject'], axis=1, inplace=True)
ds_score_df['Score'] = ds_score_df['Score'].replace(to_replace=np.nan, value=0)
ds_score_df.rename(columns={'Score':'DS'}, inplace=True)
ds_score_df.drop(['Subject'], axis=1, inplace=True)
```

```python
new_df = physics_score_df.merge(math_score_df, on=['ID','Age','Sex'], how='inner').merge(ds_score_df,on=['ID','Age','Sex'], how='inner')
new_df['Sex'] = np.where(new_df['Sex'] == 'M', 1,2)
new_df.to_csv('../requirement/ScoreFinal.csv',index=False,columns=new_df[['ID','Age','Sex','Physics','Maths','DS']])
```
