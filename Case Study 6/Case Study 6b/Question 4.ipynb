#### Create csv file from the data below and read it in the pandas’s dataframe.

#### Phase 1: Reading data

```python
import pandas as pd
sales_df = pd.read_csv('../../Case Study 7/requirements/sample-salesv2.csv')
sales_df.info()
```

#### Phase 2: Describe the data
##### Describe the data on the unitprice.

```python
sales_df['unit price'].describe()
```

#### Phase 3: Filter the data
##### Create  a new  dataframe  having  columns 'name','net_price','date'  and  group  all  the records according to name.

```python
new_sales_df = sales_df[['name','net_price','date']]
new_sales_df = new_sales_df.groupby('name',as_index = False)
```

#### Phase 4: Plotting graph
##### Plot the graph after calculating total sales by each customer. Customer name should be the onthe  x-axisand  total  saleson  y-axis.Refer  to  sample-salesv2  to  answer  the following questions

```python
import matplotlib.pyplot as plt

agg_df = new_sales_df['net_price'].sum()

plt.figure(figsize = (17,6))
plt.bar(agg_df['name'], agg_df['net_price'],color='DarkViolet')
plt.xticks(rotation = 90)
plt.xlabel('Customer Name')
plt.ylabel('Net Price')
plt.title('Net Price by Customer')
plt.show()
```
