#### Draw a scatter graph of any 50 random values of the x and y-axis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.DataFrame({'x': pd.Series(np.random.rand(50)),'y':pd.Series(np.random.randn(50))})
plt.scatter(df.x,df.y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Scatter Plot of 50 Random Values')
plt.show()
```

#### Create a dataframe from the following data

```python
data = \
    {
    'first_name': ['Jason', 'Molly', 'Tina', 'Jake', 'Amy'],
    'last_name': ['Miller', 'Jacobson', 'Ali', 'Milner', 'Cooze'],
    'female': [0, 1, 1, 0, 1],'age': [42, 52, 36, 24, 73],
    'preTestScore': [4, 24, 31, 2, 3],
    'postTestScore': [25, 94, 57, 62, 70]
 }
```

```python
df = pd.DataFrame(data)
df
```

#### Draw a Scatter plot of preTestScore and postTestScore, with the size of each point determined by age.

```python
import matplotlib.pyplot as plt
plt.scatter(df.preTestScore,df.postTestScore,s=df.age)
plt.show()
```

#### Draw a Scatterplot from the data in question 9 of preTestScore and postTestScore with the size = 300 and the color determined by sex

```python
import matplotlib.pyplot as plt

plt.scatter(df.preTestScore,df.postTestScore,s=300,edgecolors='face',c=df['female'].astype('category'))

#alternate way
#plt.scatter(df.preTestScore,df.postTestScore,s=300,edgecolors='face',c=np.where(df.female == 0, "green", "blue"))

plt.show()
```
