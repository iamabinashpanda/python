#### You are given a dataset, which is present in the LMS, containing the number of hurricanes occurring in the United States along the coast of the Atlantic. Load the data from the dataset into your program and plot a Bar Graph of the data, taking the Year as the x-axis and the number of hurricanes occurring as the Y-axis.

```python
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d.art3d import rotate_axes
from sympy import rotations

hurr_data = pd.read_csv('../../Case Study 7/requirements/Hurricanes.csv')
plt.figure(figsize=(16,5))
plt.bar(hurr_data.Year,hurr_data.Hurricanes,color='cyan')
plt.style.use('ggplot')
plt.title('Hurricanes')
plt.xlabel('Year')
plt.ylabel('Hurricanes')
plt.xticks(hurr_data.Year,rotation=90)
plt.legend(labels=['Hurricanes'])
plt.show()
```

#### The dataset given, records data of city temperatures over the years 2014 and 2015. Plot the histogram of the temperatures over this period for the cities of San Francisco and Moscow

```python
import pandas as pd
import matplotlib.pyplot as plt
city_temp_data = pd.read_csv('../../Case Study 7/requirements/CityTemps.csv')
df_melbourne = city_temp_data['Melbourne']
df_san_francisco = city_temp_data['San Francisco']
plt.figure(figsize=(16,5))
plt.hist(df_melbourne,len(df_melbourne),color='yellow',label='Melbourne')
plt.hist(df_san_francisco,len(df_san_francisco),color='orange',label='San Francisco')
plt.title('Temperature over Period')
plt.xlabel('Year')
plt.ylabel('Temperature')
plt.legend()
plt.show()
plt.show()
```

#### Plot a pie chart of the number of models released by every manufacturer, recorded in the data provided. Also mention the name of the manufacturer with the largest releases.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
cars_data = pd.read_csv('../../Case Study 7/requirements/Cars2015.csv')

#data cleansing
cars_data['Make'] = cars_data['Make'].str.strip()
cars_data['Model'] = cars_data['Model'].str.strip()

#group by
df = dict()
for group,data in cars_data.groupby("Make"):
    df[group] = len(data.Model)

#for use of explode attr n pie to show the largest manufacturer in chart
exploding_obj = np.zeros(len(df.values()))
exploding_obj[list(df.values()).index(max(df.values()))]=1

#pie chart
plt.figure(figsize=(16,8))
plt.pie(df.values(),labels=df.keys(),autopct='%d',rotatelabels=True,explode=exploding_obj,startangle=94)
plt.suptitle('Cars Models Released by Manufacturer')
plt.show()
print(f'Largest Manufacturer : {"".join([key for key, value in df.items() if value == max(df.values())])}')
```
