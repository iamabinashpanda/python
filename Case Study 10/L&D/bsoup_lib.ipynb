```python
#for edureaka links extraction
import requests
from bs4 import BeautifulSoup, BeautifulStoneSoup
from scrapy.utils.versions import scrapy_components_versions

links = []

r = requests.get('https://www.edureka.co/')
c = r.content
soup = BeautifulSoup(c,"html.parser")
soup.prettify()
for link in soup.find_all("a"):
   links.append(link.get('href'))
links = [link for link in links if link.startswith("https://")]
print(links)
```

```python
# for goodreads links extraction
import requests
from bs4 import BeautifulSoup

r1 = requests.get('https://www.goodreads.com/',headers={'User-Agent':'Mozilla/5.0'})
if(r1.status_code == 200):
    soup1 = BeautifulSoup(r1.content,"html.parser")
    categories = soup1.find_all('a',attrs={'class':'gr-hyperlink'})
    txt_categories = list()
    for category in categories:
        if 'genres' in category.get('href'):
            txt_categories.append(category.string)
    for text in txt_categories:
        print(text)
else:
    print("falied:",r1.status_code)
```
