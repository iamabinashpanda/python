```python
import pandas as pd
from bokeh.plotting import figure, show
from bokeh.io import output_notebook
from bokeh.layouts import row
```

```python
df = pd.read_excel('./verlegenhuken.xlsx')
df
```

```python
a = figure()
a.line(x=df['Temperature'],y=df['Pressure'])
```

```python
b = figure()
b.line(x=df['Year'],y=df['Pressure'])
```

```python
output_notebook()
show(row(a,b))
```
