Display **heatmap** for Divvy Bikes. Divvy Bikes runs bike rental  services in  Chicago,and  their  bike  station  geolocation  data  is  shared. Refer to BikesHeatMap.py,Divvy_Stations_2016_Q3.csv,andDivvy_Stations_2016_Q4.csv

```python
import pandas as pd
import numpy as np
import folium
```

```python
q3s_url = "https://media.githubusercontent.com/media/iamabinashpanda/python-certification-projects/refs/heads/main/Case%20Study%209/requirements/Divvy/Divvy_Stations_2016_Q3.csv?token=AUFMPTGVBGCLSBDCH4LFYRTJMSOSO"
q4s_url = "https://media.githubusercontent.com/media/iamabinashpanda/python-certification-projects/refs/heads/main/Case%20Study%209/requirements/Divvy/Divvy_Stations_2016_Q4.csv?token=AUFMPTB577EVBVWXJNB54TTJMSOU2"

q3s_df =pd.read_csv(q3s_url)
q4s_df = pd.read_csv(q4s_url)

# Variables:

# id: ID attached to each station
# name: station name
# latitude: station latitude
# longitude: station longitude
# dpcapacity: number of total docks at each station as of 9/30/2016
# online_date: date the station went live in the system
```

```python
q3s_df.head(3)
```

```python
map = folium.Map(location=[q3s_df['latitude'].mean(),q3s_df['longitude'].mean()],zoom_start=11)

feature_q3 = folium.FeatureGroup(name="Q3")
feature_q4 = folium.FeatureGroup(name="Q4")

head_data_q3 = [[row['latitude'],row['longitude']] for index, row in q3s_df.iterrows()]
head_data_q4 = [[row['latitude'],row['longitude']] for index, row in q4s_df.iterrows()]

feature_q3.add_child(folium.plugins.HeatMap(head_data_q3,blur=50))
feature_q4.add_child(folium.plugins.HeatMap(head_data_q4,blur=50))

map.add_child(feature_q3)
map.add_child(feature_q4)
map.add_child(folium.LayerControl())

map
```
