```python
import folium as f
import pandas as pd
coordinates = [
    {"loc":[38.2,-99.1],"label":"Marker A"},
    {"loc":[33.58,-98.09],"label":"Marker B"}
]
```

```python
df = pd.DataFrame(coordinates)
df[['lat','long']] = df['loc'][0],df['loc'][1]
```

```python
m = f.Map(location=[df.lat.mean(),df.long.mean()],zoom_start=7)
```

```python
for cordinate in coordinates:
    marker = f.Marker(location=cordinate["loc"],popup=cordinate["label"],icon=f.Icon(color="red"))
    marker.add_to(m)
```
