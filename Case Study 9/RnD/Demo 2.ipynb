```python
import pandas as pd
```

```python
data = pd.read_csv('../requirements/volcano_db.csv', encoding='latin-1')
data
```

```python
lon = list(data['Longitude'])
lat = list(data['Latitude'])
elev = list(data['Elev'])
```

```python
def color(elevation):
    if elevation < 1000:
        return 'green'
    elif 1000 <= elevation < 3000:
        return 'orange'
    else:
        return 'red'
```

```python
import folium
import numpy as np
```

```python
map = folium.Map(location=[np.mean(lat),np.mean(lon)],zoom_start=2)

fg_v = folium.FeatureGroup(name='Volcano')
for lt,ln,el in zip(lat,lon,elev):
    fg_v.add_child(folium.Marker(location=[lt,ln],popup=str(el),icon=folium.Icon(color=color(el),icon="home",prefix='fa')))

fg_p = folium.FeatureGroup(name='Population')
fg_p.add_child(folium.GeoJson(data=(open('../requirements/world.json', 'r', encoding='utf-8-sig').read()),
                              style_function= lambda feature: {
                                  'color': 'green' if feature['geometry']["type"]== "MultiPolygon"
                                  else 'orange' if feature['geometry']["type"]== "Polygon"
                                  else 'red'
                              }
                              ))

map.add_child(fg_v)
map.add_child(fg_p)
map.add_child(folium.LayerControl())

map
```
